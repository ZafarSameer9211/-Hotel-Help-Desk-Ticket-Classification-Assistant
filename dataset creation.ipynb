```python

# Store all dataset records here
rows = []

# Define important words for summary checking
required_words = {
    "Booking_Issue": "booking",
    "Cancellation_Issue": "cancel",
    "Checkin_Checkout": "check",
    "Room_Cleanliness": "clean",
    "Room_Amenities": "room",
    "Maintenance_Issue": "repair",
    "WiFi_Issue": "wifi",
    "Food_Service": "food",
    "Billing_Payment": "payment",
    "Customer_Service": "staff"
}

# Start ticket ID numbering
ticket_number = 1

# Loop through every category and its tickets
for category, tickets in hotel_tickets.items():

    # Loop through individual tickets
    for index, ticket_text in enumerate(tickets):

        # Assign development or evaluation split
        if index < 8:
            split = "development"
        else:
            split = "eval"

        # Create one dataset record
        rows.append({
            "id": f"H{ticket_number:03d}",
            "ticket": ticket_text,
            "expected_category": category,
            "split": split,
            "required_terms": required_words[category]
        })

        # Increase ticket ID
        ticket_number += 1

# Convert records into a Pandas DataFrame
df = pd.DataFrame(rows)

print("Dataset created successfully!")
print("Total tickets:", len(df))
display(df.head(10))
```

```python

# Check total dataset records
print("Total tickets:", df.shape[0])

# Check number of categories
print("Total categories:", df["expected_category"].nunique())

# Count tickets in each category
print("\nTickets per category:")
display(df["expected_category"].value_counts())

# Check development and evaluation records
print("\nDataset split:")
display(df["split"].value_counts())

# Check missing values
print("\nMissing values:")
display(df.isnull().sum())

# Check duplicate ticket text
print("\nDuplicate tickets:", df["ticket"].duplicated().sum())
```

```python

# Create dataset folder if it does not exist
Path("dataset").mkdir(exist_ok=True)

# Save hotel dataset inside dataset folder
df.to_csv("dataset/hotel_support_tickets.csv", index=False)

print("Hotel dataset saved successfully!")
print("File: dataset/hotel_support_tickets.csv")
print("Total records saved:", len(df))
```
